```python
# Join array using the concatinate fun
import numpy as np
n1 = np.array([2,13,15])
n2 = np.array([2,11,19])
print ("iterating array")
for a in  n1:
  print(a)
  print("iterating array n2")
  for a  in n2:
    print (a)
    resume= np.concatenate((n1,n2))
    print ("After join the array",resume)
```

```python
# Join array using the concatinate fun
import numpy as np
n1 = np.array([2,13,15])
n2 = np.array([2,11,19])
print ("iterating array")
for a in  n1:
  print(a)
  print("iterating array n2")
  for a  in n2:
    print (a)
    resume= np.stack((n1,n2),axis =1)

    print ("After join the array",resume)
```
