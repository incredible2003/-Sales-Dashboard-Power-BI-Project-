```python
import matplotlib.pyplot as plt
import numpy as np

Cricket_Viewership = np.array([80, 70, 60, 50, 40, 30, 20])
TeamName = np.array(["CSK", "RCB", "MI", "KKR", "LSG", "RR", "GT"])

plt.pie(Cricket_Viewership, labels=TeamName, autopct='%1.2f%%')
plt.title("Cricket Fanbase")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

Cricket_Viewership = np.array([80, 70, 60, 50, 40, 30, 20])
TeamName = np.array(["CSK", "RCB", "MI", "KKR", "LSG", "RR", "GT"])

# Har team ka color
colors = [
    "yellow",   # CSK
    "red",      # RCB
    "blue",     # MI
    "purple",   # KKR
    "cyan",     # LSG
    "pink",     # RR
    "skyblue"   # GT
]

plt.pie(Cricket_Viewership, labels=TeamName, autopct='%1.2f%%', colors=colors)
plt.title("Most Fanbase Tata IPL  2026 ")
plt.show()
```
