```python
# group of data
import pandas as pd
data = {
    'Player': ['MS Dhoni','Virat ', 'Rohit', 'Rinku','Shivam','Ritu raj'],
    'Rank': [1,2, 3,5,4,6],
    'year':[2021,2023,2024,2025,2008,2009]


}
df = pd.DataFrame(data)
print("cricket payer record \n",df)
groupRes = df.groupby('Player')
for name ,group in groupRes:
  print ("\n",name)
  print(group)
```

```python
# to find the mean medium and mode
import pandas as pd
import numpy as np
data = {
    'Player': ['MS Dhoni','Virat ', 'Rohit', 'Rinku','Shivam','Ritu raj'],
    'Points':[24,12,14,33,44,55],
    'Rank': [1,2, 3,5,4,6],
    'year':[2021,2023,2024,2025,2024,2021]


}
df = pd.DataFrame(data)
print("cricket player record \n",df)
groupRes = df.groupby('year')
print("\n",groupRes['Points'].agg(np.mean))
```

```python
import pandas as pd

data = {
    "Marks": [45, 78, 90, 78, 85]
}

df = pd.DataFrame(data)

# Mean
print("Mean\n", df.mean())

# Median
print("Median\n", df.median())

# Mode
print("Mode\n", df.mode())

# Standard Deviation
print("Standard Deviation\n", df.std())
# Desribe fun use
print("\n summary of data fun = \n ",df.describe())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Temperature': [14,20,34,45,56,67,78,87,88,99],
    'Humidity': [32,34,13,34,43,24,23,34,45,55],
    'Wind': [12,14,24,26,34,45,34,56,56,88],
    'Precipitation': [14,34,38,56,66,99,77,55,67,77]
}

df = pd.DataFrame(data)

# plot graph
df.plot()

# title and labels
plt.title("Weather Data Analysis")
plt.xlabel("Index")
plt.ylabel("Values")

plt.show()
```
